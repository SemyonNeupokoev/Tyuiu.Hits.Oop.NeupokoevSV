```
public class Subscription
{
    private int _subscriptionId;
    private string _serviceName;
    private decimal _cost;

    public int SubscriptionId
    {
        get { return _subscriptionId; }
        set { _subscriptionId = value; }
    }

    public string ServiceName
    {
        get { return _serviceName; }
        set { _serviceName = value; }
    }

    public decimal Cost
    {
        get { return _cost; }
        set { _cost = value; }
    }

    public Subscription(int subscriptionId, string serviceName, decimal cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Название: {ServiceName}, Стоимость: {Cost}");
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual void ExtendSubscription(int months)
    {
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Стоимость: {Cost * months}");
    }
}

public class OnlineServiceSubscription : Subscription
{
    private int _maxUsers;

    public int MaxUsers
    {
        get { return _maxUsers; }
        set { _maxUsers = value; }
    }

    public OnlineServiceSubscription(int subscriptionId, string serviceName, decimal cost, int maxUsers)
        : base(subscriptionId, serviceName, cost)
    {
        MaxUsers = maxUsers;
    }

    public override decimal CalculateMonthlyCost()
    {
        return Cost + (MaxUsers - 1) * Cost * 0.1m;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Пользователей: {MaxUsers}");
    }
}

public class StreamingSubscription : Subscription
{
    private int _maxStreams;

    public int MaxStreams
    {
        get { return _maxStreams; }
        set { _maxStreams = value; }
    }

    public StreamingSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost)
    {
        MaxStreams = maxStreams;
    }

    public override void ExtendSubscription(int months)
    {
        decimal discount = 0;
        if (months >= 12) discount = 0.2m;
        else if (months >= 6) discount = 0.1m;

        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Скидка: {discount * 100}%");
        Console.WriteLine($"Итоговая стоимость: {Cost * months * (1 - discount)}");
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Потоков: {MaxStreams}");
    }
}

public class User
{
    private string _name;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public User(string name)
    {
        Name = name;
    }

    public void ShowSubscription(Subscription sub)
    {
        Console.WriteLine($"Пользователь {Name} смотрит подписку:");
        sub.GetSubscriptionDetails();
        Console.WriteLine($"Ежемесячная стоимость: {sub.CalculateMonthlyCost()}");
    }
}

User user = new User("Иван");

OnlineServiceSubscription online = new OnlineServiceSubscription(1, "Microsoft 365", 500, 5);
StreamingSubscription stream = new StreamingSubscription(2, "Netflix", 899, 4);

user.ShowSubscription(online);
Console.WriteLine();
user.ShowSubscription(stream);

Console.WriteLine();
stream.ExtendSubscription(12);
```
