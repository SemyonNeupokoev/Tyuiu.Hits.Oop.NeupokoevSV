```
public interface IInstallable
{
    void Install();
    void Uninstall();
}

public interface IComponentRepository
{
    void Add(Component c);
    bool Remove(Component c);
    IEnumerable<Component> GetAll();
    Component FindByName(string name);
}

public class InMemoryComponentRepository : IComponentRepository
{
    private readonly List<Component> _items = new();

    public void Add(Component c) => _items.Add(c);
    public bool Remove(Component c) => _items.Remove(c);
    public IEnumerable<Component> GetAll() => _items;
    public Component FindByName(string name) =>
        _items.Find(x => x.Name == name);
}

public abstract class Component
{
    private string _name;
    private string _manufacturer;
    private decimal _price;
    private int _powerConsumption;
    private string _serialNumber;
    private int _warrantyMonths;
    private DateTime _purchaseDate;
    private bool _isInstalled;

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым.");
            _name = value;
        }
    }

    public string Manufacturer
    {
        get => _manufacturer;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Производитель не может быть пустым.");
            _manufacturer = value;
        }
    }

    public decimal Price
    {
        get => _price;
        set
        {
            if (value < 0) throw new ArgumentException("Цена не может быть отрицательной.");
            _price = value;
        }
    }

    public int PowerConsumption
    {
        get => _powerConsumption;
        set
        {
            if (value < 0) throw new ArgumentException("Потребление не может быть отрицательным.");
            _powerConsumption = value;
        }
    }

    public string SerialNumber
    {
        get => _serialNumber;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Серийный номер обязателен.");
            _serialNumber = value;
        }
    }

    public int WarrantyMonths
    {
        get => _warrantyMonths;
        set
        {
            if (value < 0) throw new ArgumentException("Гарантия не может быть отрицательной.");
            _warrantyMonths = value;
        }
    }

    public DateTime PurchaseDate
    {
        get => _purchaseDate;
        set => _purchaseDate = value;
    }

    public bool IsInstalled
    {
        get => _isInstalled;
        protected set => _isInstalled = value;
    }

    protected Component(string name, string manufacturer, decimal price, int powerConsumption,
                        string serialNumber, int warrantyMonths, DateTime purchaseDate)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
        PowerConsumption = powerConsumption;
        SerialNumber = serialNumber;
        WarrantyMonths = warrantyMonths;
        PurchaseDate = purchaseDate;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Производитель: {Manufacturer}, Цена: {Price}, Потребление: {PowerConsumption} Вт");
        Console.WriteLine($"SN: {SerialNumber}, Гарантия: {WarrantyMonths} мес., Куплен: {PurchaseDate:dd.MM.yyyy}, Установлен: {IsInstalled}");
    }

    public virtual bool IsUnderWarranty() =>
        DateTime.Now < PurchaseDate.AddMonths(WarrantyMonths);

    public virtual decimal GetDepreciation()
    {
        int monthsUsed = Math.Max(0, (DateTime.Now.Year - PurchaseDate.Year) * 12
                                     + DateTime.Now.Month - PurchaseDate.Month);
        decimal depreciation = Price * 0.02m * monthsUsed;
        return Math.Max(0, Price - depreciation);
    }

    public virtual string GetStatus() => IsInstalled ? "Установлен" : "Не установлен";

    public override string ToString() => $"{Name} ({Manufacturer}) - {Price} руб.";

    public static bool operator ==(Component a, Component b)
    {
        if (a is null && b is null) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(Component a, Component b) => !(a == b);

    public override bool Equals(object obj) => obj is Component c && this == c;

    public override int GetHashCode() => HashCode.Combine(Name, Manufacturer, Price);
}

public class CPU : Component, IInstallable
{
    private int _cores;
    private double _frequency;
    private string _socket;

    public int Cores
    {
        get => _cores;
        set
        {
            if (value <= 0) throw new ArgumentException("Ядер должно быть > 0.");
            _cores = value;
        }
    }

    public double Frequency
    {
        get => _frequency;
        set
        {
            if (value <= 0) throw new ArgumentException("Частота должна быть > 0.");
            _frequency = value;
        }
    }

    public string Socket
    {
        get => _socket;
        set => _socket = value;
    }

    public CPU(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int cores, double frequency, string socket)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        Cores = cores;
        Frequency = frequency;
        Socket = socket;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Ядер: {Cores}, Частота: {Frequency} ГГц, Сокет: {Socket}");
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine($"CPU {Name} установлен в сокет {Socket}.");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine($"CPU {Name} извлечён из сокета {Socket}.");
    }

    public override string GetStatus() =>
        $"[CPU] {Name}: {Cores} ядер @ {Frequency} ГГц — {(IsInstalled ? "установлен" : "свободен")}";
}

public class RAM : Component, IInstallable
{
    private int _capacityGB;
    private int _frequencyMHz;

    public int CapacityGB
    {
        get => _capacityGB;
        set
        {
            if (value <= 0) throw new ArgumentException("Объём > 0.");
            _capacityGB = value;
        }
    }

    public int FrequencyMHz
    {
        get => _frequencyMHz;
        set
        {
            if (value <= 0) throw new ArgumentException("Частота > 0.");
            _frequencyMHz = value;
        }
    }

    public RAM(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int capacityGB, int frequencyMHz)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        CapacityGB = capacityGB;
        FrequencyMHz = frequencyMHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объём: {CapacityGB} ГБ, Частота: {FrequencyMHz} МГц");
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine($"RAM {Name} установлена в слот DIMM.");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine($"RAM {Name} извлечена из слота.");
    }
}

public class GPU : Component, IInstallable
{
    private int _videoMemoryGB;
    private string _interface;

    public int VideoMemoryGB
    {
        get => _videoMemoryGB;
        set
        {
            if (value <= 0) throw new ArgumentException("Видеопамять > 0.");
            _videoMemoryGB = value;
        }
    }

    public string Interface
    {
        get => _interface;
        set => _interface = value;
    }

    public GPU(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int videoMemoryGB, string @interface)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        VideoMemoryGB = videoMemoryGB;
        Interface = @interface;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Видеопамять: {VideoMemoryGB} ГБ, Интерфейс: {Interface}");
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine($"GPU {Name} установлена в слот {Interface}.");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine($"GPU {Name} извлечена из слота {Interface}.");
    }
}

public class HDD : Component, IInstallable
{
    private int _capacityGB;
    private int _rpm;

    public int CapacityGB
    {
        get => _capacityGB;
        set
        {
            if (value <= 0) throw new ArgumentException("Объём > 0.");
            _capacityGB = value;
        }
    }

    public int Rpm
    {
        get => _rpm;
        set => _rpm = value;
    }

    public HDD(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int capacityGB, int rpm)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        CapacityGB = capacityGB;
        Rpm = rpm;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объём: {CapacityGB} ГБ, Скорость: {Rpm} об/мин");
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine($"HDD {Name} установлен в отсек 3.5\".");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine($"HDD {Name} извлечён из отсека.");
    }
}

public class Computer
{
    private string _model;
    private readonly List<Component> _components = new();

    public string Model
    {
        get => _model;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Модель не может быть пустой.");
            _model = value;
        }
    }

    public Computer(string model)
    {
        Model = model;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        ((IInstallable)component).Install();
        Console.WriteLine($"Компонент {component.Name} добавлен в {Model}.");
    }

    public void RemoveComponent<T>(T component) where T : Component, IInstallable
    {
        if (_components.Remove(component))
        {
            ((IInstallable)component).Uninstall();
            Console.WriteLine($"Компонент {component.Name} удалён из {Model}.");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден в {Model}.");
        }
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var c in _components) total += c.Price;
        return total;
    }

    public void ShowAll()
    {
        Console.WriteLine($"--- Компьютер {Model} ({_components.Count} компонентов) ---");
        foreach (var c in _components)
        {
            c.GetInfo();
            Console.WriteLine($"Статус: {c.GetStatus()}");
            Console.WriteLine($"На гарантии: {(c.IsUnderWarranty() ? "да" : "нет")}, Остаточная стоимость: {c.GetDepreciation():F2} руб.");
            Console.WriteLine();
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.");
    }
}

public interface IComputerBuilder
{
    Computer Build(string model);
}

public class GamingComputerBuilder : IComputerBuilder
{
    public Computer Build(string model)
    {
        var pc = new Computer(model);
        var today = DateTime.Now;
        pc.AddComponent(new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, "CPU-7800X3D", 36, today, 8, 4.2, "AM5"));
        pc.AddComponent(new RAM("DDR5 32GB", "Kingston", 12000, 15, "RAM-DDR5-32", 24, today, 32, 6000));
        pc.AddComponent(new GPU("RTX 4080", "NVIDIA", 120000, 320, "GPU-RTX4080", 36, today, 16, "PCIe 4.0"));
        pc.AddComponent(new HDD("SSD 1TB", "Samsung", 10000, 6, "SSD-1TB", 60, today, 1000, 7200));
        return pc;
    }
}

public class OfficeComputerBuilder : IComputerBuilder
{
    public Computer Build(string model)
    {
        var pc = new Computer(model);
        var today = DateTime.Now;
        pc.AddComponent(new CPU("Core i5-12400", "Intel", 18000, 65, "CPU-i5", 24, today, 6, 2.5, "LGA1700"));
        pc.AddComponent(new RAM("DDR4 16GB", "Crucial", 5000, 10, "RAM-DDR4-16", 24, today, 16, 3200));
        pc.AddComponent(new HDD("SSD 512GB", "WD", 4500, 4, "SSD-512", 36, today, 512, 7200));
        return pc;
    }
}

public class ComputerService
{
    private readonly IComputerBuilder _builder;

    public ComputerService(IComputerBuilder builder)
    {
        _builder = builder ?? throw new ArgumentNullException(nameof(builder));
    }

    public Computer Assemble(string model) => _builder.Build(model);
}

Console.WriteLine("=== Игровой ПК ===");
var gamingService = new ComputerService(new GamingComputerBuilder());
var gaming = gamingService.Assemble("Gaming PC");
gaming.ShowAll();

Console.WriteLine("=== Офисный ПК ===");
var officeService = new ComputerService(new OfficeComputerBuilder());
var office = officeService.Assemble("Office PC");
office.ShowAll();

Console.WriteLine("=== Сравнение CPU ===");
var today2 = DateTime.Now;
var cpu1 = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, "SN1", 36, today2, 8, 4.2, "AM5");
var cpu2 = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, "SN2", 36, today2, 8, 4.2, "AM5");
var cpu3 = new CPU("Core i5-12400", "Intel", 18000, 65, "SN3", 24, today2, 6, 2.5, "LGA1700");
Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 != cpu3: {cpu1 != cpu3}");
Console.WriteLine(cpu1.ToString());
Console.WriteLine($"На гарантии: {cpu1.IsUnderWarranty()}");
Console.WriteLine($"Остаточная стоимость: {cpu1.GetDepreciation():F2} руб.");
```
