<h3 style="color:DodgerBlue">Описание задачи:</h3>
Создать базовый класс Subscription в C#, который будет представлять подписки на
различные услуги. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

<h3 style="color:DodgerBlue">Требования к базовому классу Subscription:</h3>

• Атрибуты: ID подписки (SubscriptionId), Название услуги (ServiceName),
Стоимость подписки (Cost).
• Методы:
    o CalculateMonthlyCost(): метод для расчета ежемесячной стоимости
    подписки.
    o ExtendSubscription(): метод для продления подписки на
    дополнительный период.
    o GetSubscriptionDetails(): метод для получения деталей подписки.

Требования к производным классам:
1. ПодпискаНаОнлайнСервис (OnlineServiceSubscription): 
Должна содержать дополнительные атрибуты, такие как Количество доступных пользователей
(MaxUsers). Метод CalculateMonthlyCost() должен быть переопределен для
учета количества пользователей при расчете стоимости.


2. ПодпискаНаСтreamинг (StreamingSubscription): Должна содержать
дополнительные атрибуты, такие как Количество одновременных потоков
(MaxStreams). Метод ExtendSubscription() должен быть переопределен для
добавления специальных предложений для продления подписки.

3. ПодпискаНаВидео(VideoSubscription) (если требуется третий класс): Должна
содержать дополнительные атрибуты, такие как Качество видео
(VideoQuality). Метод GetSubscriptionDetails() должен быть переопределен
для отображения качества видео вместе с другими деталями подписки.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Subscription
{
    public int SubscriptionId;
    public string ServiceName;
    public decimal Cost;

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Название: {ServiceName}, Стоимость: {Cost}");
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual void ExtendSubscription(int months)
    {
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Стоимость: {Cost * months}");
    }
}

public class OnlineServiceSubscription : Subscription
{
    public int MaxUsers;

    public override decimal CalculateMonthlyCost()
    {
        return Cost + (MaxUsers - 1) * Cost * 0.1m;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Пользователей: {MaxUsers}");
    }
}

public class StreamingSubscription : Subscription
{
    public int MaxStreams;

    public override void ExtendSubscription(int months)
    {
        decimal discount = 0;
        if (months >= 12) discount = 0.2m;
        else if (months >= 6) discount = 0.1m;

        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Скидка: {discount * 100}%");
        Console.WriteLine($"Итоговая стоимость: {Cost * months * (1 - discount)}");
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Потоков: {MaxStreams}");
    }
}

OnlineServiceSubscription online = new OnlineServiceSubscription();
online.SubscriptionId = 1;
online.ServiceName = "Microsoft 365";
online.Cost = 500;
online.MaxUsers = 5;
online.GetSubscriptionDetails();
Console.WriteLine($"Стоимость: {online.CalculateMonthlyCost()}");
Console.WriteLine();

StreamingSubscription stream = new StreamingSubscription();
stream.SubscriptionId = 2;
stream.ServiceName = "Netflix";
stream.Cost = 799;
stream.MaxStreams = 4;
stream.GetSubscriptionDetails();
stream.ExtendSubscription(12);
```
