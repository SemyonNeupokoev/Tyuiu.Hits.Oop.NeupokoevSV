```
public class Subscription
{
    private int _subscriptionId;
    private string _serviceName;
    private decimal _cost;
    private bool _isActive;

    public int SubscriptionId
    {
        get { return _subscriptionId; }
        set { _subscriptionId = value; }
    }

    public string ServiceName
    {
        get { return _serviceName; }
        set { _serviceName = value; }
    }

    public decimal Cost
    {
        get { return _cost; }
        set { _cost = value; }
    }

    public bool IsActive
    {
        get { return _isActive; }
        set { _isActive = value; }
    }

    public Subscription(int subscriptionId, string serviceName, decimal cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
        IsActive = true;
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Название: {ServiceName}, Стоимость: {Cost}, Активна: {IsActive}");
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual void ExtendSubscription(int months)
    {
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Стоимость: {Cost * months}");
    }

    public void Activate()
    {
        IsActive = true;
        Console.WriteLine($"Подписка {ServiceName} активирована.");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"Подписка {ServiceName} деактивирована.");
    }
}

public class OnlineServiceSubscription : Subscription
{
    private int _maxUsers;
    private int _storageGB;

    public int MaxUsers
    {
        get { return _maxUsers; }
        set { _maxUsers = value; }
    }

    public int StorageGB
    {
        get { return _storageGB; }
        set { _storageGB = value; }
    }

    public OnlineServiceSubscription(int subscriptionId, string serviceName, decimal cost, int maxUsers)
        : base(subscriptionId, serviceName, cost)
    {
        MaxUsers = maxUsers;
        StorageGB = 10;
    }

    public override decimal CalculateMonthlyCost()
    {
        return Cost + (MaxUsers - 1) * Cost * 0.1m;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Пользователей: {MaxUsers}, Хранилище: {StorageGB} ГБ");
    }

    public void UpgradeStorage(int extraGB)
    {
        StorageGB += extraGB;
        Console.WriteLine($"Хранилище увеличено на {extraGB} ГБ. Теперь: {StorageGB} ГБ");
    }
}

public class StreamingSubscription : Subscription
{
    private int _maxStreams;
    private string _videoQuality;

    public int MaxStreams
    {
        get { return _maxStreams; }
        set { _maxStreams = value; }
    }

    public string VideoQuality
    {
        get { return _videoQuality; }
        set { _videoQuality = value; }
    }

    public StreamingSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost)
    {
        MaxStreams = maxStreams;
        VideoQuality = "HD";
    }

    public override void ExtendSubscription(int months)
    {
        decimal discount = 0;
        if (months >= 12) discount = 0.2m;
        else if (months >= 6) discount = 0.1m;

        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Скидка: {discount * 100}%");
        Console.WriteLine($"Итоговая стоимость: {Cost * months * (1 - discount)}");
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Потоков: {MaxStreams}, Качество: {VideoQuality}");
    }

    public void UpgradeQuality(string newQuality)
    {
        VideoQuality = newQuality;
        Console.WriteLine($"Качество видео изменено на: {VideoQuality}");
    }
}

public class PremiumStreamingSubscription : StreamingSubscription
{
    private int _exclusiveContentCount;

    public int ExclusiveContentCount
    {
        get { return _exclusiveContentCount; }
        set { _exclusiveContentCount = value; }
    }

    public PremiumStreamingSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost, maxStreams)
    {
        ExclusiveContentCount = 50;
        VideoQuality = "4K";
    }

    public override decimal CalculateMonthlyCost()
    {
        return base.CalculateMonthlyCost() + 300;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Эксклюзивного контента: {ExclusiveContentCount}");
    }

    public void AddExclusiveContent(int count)
    {
        ExclusiveContentCount += count;
        Console.WriteLine($"Добавлено {count} единиц контента. Всего: {ExclusiveContentCount}");
    }
}

public interface IDiscountable
{
    decimal ApplyDiscount(decimal amount, int percent);
}

public interface IGiftable
{
    void GiftTo(string recipient);
}

public class GiftablePremiumSubscription : PremiumStreamingSubscription, IDiscountable, IGiftable
{
    public GiftablePremiumSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost, maxStreams)
    {
    }

    public decimal ApplyDiscount(decimal amount, int percent)
    {
        decimal result = amount * (1 - percent / 100m);
        Console.WriteLine($"Применена скидка {percent}%: {amount} -> {result}");
        return result;
    }

    public void GiftTo(string recipient)
    {
        Console.WriteLine($"Подписка {ServiceName} подарена пользователю {recipient}");
    }
}

public class User
{
    private string _name;
    private decimal _balance;
    private bool _isPremium;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public decimal Balance
    {
        get { return _balance; }
        set { _balance = value; }
    }

    public bool IsPremium
    {
        get { return _isPremium; }
        set { _isPremium = value; }
    }

    public User(string name)
    {
        Name = name;
        Balance = 0;
        IsPremium = false;
    }

    public void ShowSubscription(Subscription sub)
    {
        Console.WriteLine($"Пользователь {Name} смотрит подписку:");
        sub.GetSubscriptionDetails();
        Console.WriteLine($"Ежемесячная стоимость: {sub.CalculateMonthlyCost()}");
    }

    public void Deposit(decimal amount)
    {
        Balance += amount;
        Console.WriteLine($"{Name} пополнил баланс на {amount}. Баланс: {Balance}");
    }

    public bool Pay(decimal amount)
    {
        if (Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"{Name} оплатил {amount}. Остаток: {Balance}");
            return true;
        }
        Console.WriteLine($"Недостаточно средств у {Name}. Нужно: {amount}, есть: {Balance}");
        return false;
    }

    public void ShowBalance()
    {
        Console.WriteLine($"Баланс {Name}: {Balance}, Премиум: {IsPremium}");
    }
}

User user = new User("Иван");
user.Deposit(2000);

OnlineServiceSubscription online = new OnlineServiceSubscription(1, "Microsoft 365", 500, 5);
StreamingSubscription stream = new StreamingSubscription(2, "Netflix", 899, 4);

Console.WriteLine("Простое наследование");
user.ShowSubscription(online);
Console.WriteLine();
user.ShowSubscription(stream);

Console.WriteLine("Сложное наследование");
PremiumStreamingSubscription premium = new PremiumStreamingSubscription(3, "Netflix Premium", 1299, 6);
user.ShowSubscription(premium);
premium.AddExclusiveContent(10);

Console.WriteLine("Множественное наследование");
GiftablePremiumSubscription gift = new GiftablePremiumSubscription(4, "Netflix Gift", 1499, 6);
user.ShowSubscription(gift);
gift.GiftTo("Мария");
gift.ApplyDiscount(1499, 15);

Console.WriteLine("Дополнительные методы");
online.UpgradeStorage(20);
stream.UpgradeQuality("4K");
stream.ExtendSubscription(12);
premium.Deactivate();
premium.Activate();

Console.WriteLine(" Оплата ");
user.Pay(1200);
user.ShowBalance();
```
