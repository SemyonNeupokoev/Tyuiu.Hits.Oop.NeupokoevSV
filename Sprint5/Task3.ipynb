```
public interface IInstallable
{
    void Install();
    void Uninstall();
}

public interface IComponentRepository
{
    void Add(Component c);
    bool Remove(Component c);
    IEnumerable<Component> GetAll();
    Component FindByName(string name);
}

public class InMemoryComponentRepository : IComponentRepository
{
    private List<Component> _items;

    public InMemoryComponentRepository()
    {
        _items = new List<Component>();
    }

    public void Add(Component c)
    {
        _items.Add(c);
    }

    public bool Remove(Component c)
    {
        return _items.Remove(c);
    }

    public IEnumerable<Component> GetAll()
    {
        return _items;
    }

    public Component FindByName(string name)
    {
        foreach (Component c in _items)
        {
            if (c.Name == name)
            {
                return c;
            }
        }
        return null;
    }
}

public abstract class Component
{
    private string _name;
    private string _manufacturer;
    private decimal _price;
    private int _powerConsumption;
    private string _serialNumber;
    private int _warrantyMonths;
    private DateTime _purchaseDate;
    private bool _isInstalled;

    public string Name
    {
        get { return _name; }
        set
        {
            if (value == null || value == "")
            {
                throw new ArgumentException("Имя не может быть пустым.");
            }
            _name = value;
        }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set
        {
            if (value == null || value == "")
            {
                throw new ArgumentException("Производитель не может быть пустым.");
            }
            _manufacturer = value;
        }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value < 0)
            {
                throw new ArgumentException("Цена не может быть отрицательной.");
            }
            _price = value;
        }
    }

    public int PowerConsumption
    {
        get { return _powerConsumption; }
        set
        {
            if (value < 0)
            {
                throw new ArgumentException("Потребление не может быть отрицательным.");
            }
            _powerConsumption = value;
        }
    }

    public string SerialNumber
    {
        get { return _serialNumber; }
        set
        {
            if (value == null || value == "")
            {
                throw new ArgumentException("Серийный номер обязателен.");
            }
            _serialNumber = value;
        }
    }

    public int WarrantyMonths
    {
        get { return _warrantyMonths; }
        set
        {
            if (value < 0)
            {
                throw new ArgumentException("Гарантия не может быть отрицательной.");
            }
            _warrantyMonths = value;
        }
    }

    public DateTime PurchaseDate
    {
        get { return _purchaseDate; }
        set { _purchaseDate = value; }
    }

    public bool IsInstalled
    {
        get { return _isInstalled; }
        protected set { _isInstalled = value; }
    }

    protected Component(string name, string manufacturer, decimal price, int powerConsumption,
                        string serialNumber, int warrantyMonths, DateTime purchaseDate)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
        PowerConsumption = powerConsumption;
        SerialNumber = serialNumber;
        WarrantyMonths = warrantyMonths;
        PurchaseDate = purchaseDate;
        IsInstalled = false;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine("Компонент: " + Name + ", Производитель: " + Manufacturer
                          + ", Цена: " + Price + ", Потребление: " + PowerConsumption + " Вт");
        Console.WriteLine("SN: " + SerialNumber + ", Гарантия: " + WarrantyMonths
                          + " мес., Куплен: " + PurchaseDate.ToString("dd.MM.yyyy")
                          + ", Установлен: " + IsInstalled);
    }

    public virtual bool IsUnderWarranty()
    {
        DateTime endDate = PurchaseDate.AddMonths(WarrantyMonths);
        if (DateTime.Now < endDate)
        {
            return true;
        }
        return false;
    }

    public virtual decimal GetDepreciation()
    {
        int monthsUsed = (DateTime.Now.Year - PurchaseDate.Year) * 12
                       + (DateTime.Now.Month - PurchaseDate.Month);
        if (monthsUsed < 0)
        {
            monthsUsed = 0;
        }

        decimal depreciation = Price * 0.02m * monthsUsed;
        decimal result = Price - depreciation;

        if (result < 0)
        {
            result = 0;
        }
        return result;
    }

    public virtual string GetStatus()
    {
        if (IsInstalled == true)
        {
            return "Установлен";
        }
        return "Не установлен";
    }

    public override string ToString()
    {
        return Name + " (" + Manufacturer + ") - " + Price + " руб.";
    }

    public static bool operator ==(Component a, Component b)
    {
        if (ReferenceEquals(a, null) && ReferenceEquals(b, null))
        {
            return true;
        }
        if (ReferenceEquals(a, null) || ReferenceEquals(b, null))
        {
            return false;
        }
        if (a.Name == b.Name && a.Price == b.Price)
        {
            return true;
        }
        return false;
    }

    public static bool operator !=(Component a, Component b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        Component c = obj as Component;
        if (c == null)
        {
            return false;
        }
        return this == c;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price);
    }
}

public class CPU : Component, IInstallable
{
    private int _cores;
    private double _frequency;
    private string _socket;

    public int Cores
    {
        get { return _cores; }
        set
        {
            if (value <= 0)
            {
                throw new ArgumentException("Ядер должно быть больше нуля.");
            }
            _cores = value;
        }
    }

    public double Frequency
    {
        get { return _frequency; }
        set
        {
            if (value <= 0)
            {
                throw new ArgumentException("Частота должна быть больше нуля.");
            }
            _frequency = value;
        }
    }

    public string Socket
    {
        get { return _socket; }
        set { _socket = value; }
    }

    public CPU(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int cores, double frequency, string socket)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        Cores = cores;
        Frequency = frequency;
        Socket = socket;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine("Ядер: " + Cores + ", Частота: " + Frequency + " ГГц, Сокет: " + Socket);
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine("CPU " + Name + " установлен в сокет " + Socket + ".");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine("CPU " + Name + " извлечён из сокета " + Socket + ".");
    }

    public override string GetStatus()
    {
        string state;
        if (IsInstalled == true)
        {
            state = "установлен";
        }
        else
        {
            state = "свободен";
        }
        return "[CPU] " + Name + ": " + Cores + " ядер @ " + Frequency + " ГГц — " + state;
    }
}

public class RAM : Component, IInstallable
{
    private int _capacityGB;
    private int _frequencyMHz;

    public int CapacityGB
    {
        get { return _capacityGB; }
        set
        {
            if (value <= 0)
            {
                throw new ArgumentException("Объём должен быть больше нуля.");
            }
            _capacityGB = value;
        }
    }

    public int FrequencyMHz
    {
        get { return _frequencyMHz; }
        set
        {
            if (value <= 0)
            {
                throw new ArgumentException("Частота должна быть больше нуля.");
            }
            _frequencyMHz = value;
        }
    }

    public RAM(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int capacityGB, int frequencyMHz)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        CapacityGB = capacityGB;
        FrequencyMHz = frequencyMHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine("Объём: " + CapacityGB + " ГБ, Частота: " + FrequencyMHz + " МГц");
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine("RAM " + Name + " установлена в слот DIMM.");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine("RAM " + Name + " извлечена из слота.");
    }
}

public class GPU : Component, IInstallable
{
    private int _videoMemoryGB;
    private string _interface;

    public int VideoMemoryGB
    {
        get { return _videoMemoryGB; }
        set
        {
            if (value <= 0)
            {
                throw new ArgumentException("Видеопамять должна быть больше нуля.");
            }
            _videoMemoryGB = value;
        }
    }

    public string Interface
    {
        get { return _interface; }
        set { _interface = value; }
    }

    public GPU(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int videoMemoryGB, string interfac)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        VideoMemoryGB = videoMemoryGB;
        Interface = interfac;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine("Видеопамять: " + VideoMemoryGB + " ГБ, Интерфейс: " + Interface);
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine("GPU " + Name + " установлена в слот " + Interface + ".");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine("GPU " + Name + " извлечена из слота " + Interface + ".");
    }
}

public class HDD : Component, IInstallable
{
    private int _capacityGB;
    private int _rpm;

    public int CapacityGB
    {
        get { return _capacityGB; }
        set
        {
            if (value <= 0)
            {
                throw new ArgumentException("Объём должен быть больше нуля.");
            }
            _capacityGB = value;
        }
    }

    public int Rpm
    {
        get { return _rpm; }
        set { _rpm = value; }
    }

    public HDD(string name, string manufacturer, decimal price, int powerConsumption,
               string serialNumber, int warrantyMonths, DateTime purchaseDate,
               int capacityGB, int rpm)
        : base(name, manufacturer, price, powerConsumption, serialNumber, warrantyMonths, purchaseDate)
    {
        CapacityGB = capacityGB;
        Rpm = rpm;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine("Объём: " + CapacityGB + " ГБ, Скорость: " + Rpm + " об/мин");
    }

    void IInstallable.Install()
    {
        IsInstalled = true;
        Console.WriteLine("HDD " + Name + " установлен в отсек 3.5\".");
    }

    void IInstallable.Uninstall()
    {
        IsInstalled = false;
        Console.WriteLine("HDD " + Name + " извлечён из отсека.");
    }
}

public class Computer
{
    private string _model;
    private List<Component> _components;

    public string Model
    {
        get { return _model; }
        set
        {
            if (value == null || value == "")
            {
                throw new ArgumentException("Модель не может быть пустой.");
            }
            _model = value;
        }
    }

    public Computer(string model)
    {
        Model = model;
        _components = new List<Component>();
    }

    public void AddComponent(Component component)
    {
        IInstallable installable = component as IInstallable;
        if (installable == null)
        {
            Console.WriteLine("Компонент " + component.Name + " нельзя установить.");
            return;
        }

        _components.Add(component);
        installable.Install();
        Console.WriteLine("Компонент " + component.Name + " добавлен в " + Model + ".");
    }

    public void RemoveComponent(Component component)
    {
        if (_components.Remove(component) == true)
        {
            IInstallable installable = component as IInstallable;
            if (installable != null)
            {
                installable.Uninstall();
            }
            Console.WriteLine("Компонент " + component.Name + " удалён из " + Model + ".");
        }
        else
        {
            Console.WriteLine("Компонент " + component.Name + " не найден в " + Model + ".");
        }
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (Component c in _components)
        {
            total = total + c.Price;
        }
        return total;
    }

    public void ShowAll()
    {
        Console.WriteLine("--- Компьютер " + Model + " (" + _components.Count + " компонентов) ---");
        foreach (Component c in _components)
        {
            c.GetInfo();
            Console.WriteLine("Статус: " + c.GetStatus());
            if (c.IsUnderWarranty() == true)
            {
                Console.WriteLine("На гарантии: да");
            }
            else
            {
                Console.WriteLine("На гарантии: нет");
            }
            Console.WriteLine("Остаточная стоимость: " + c.GetDepreciation().ToString("F2") + " руб.");
            Console.WriteLine();
        }
        Console.WriteLine("Общая стоимость: " + GetTotalPrice() + " руб.");
    }
}

public interface IComputerBuilder
{
    Computer Build(string model);
}

public class GamingComputerBuilder : IComputerBuilder
{
    public Computer Build(string model)
    {
        Computer pc = new Computer(model);
        DateTime today = DateTime.Now;

        CPU cpu = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120,
                          "CPU-7800X3D", 36, today, 8, 4.2, "AM5");
        RAM ram = new RAM("DDR5 32GB", "Kingston", 12000, 15,
                          "RAM-DDR5-32", 24, today, 32, 6000);
        GPU gpu = new GPU("RTX 4080", "NVIDIA", 120000, 320,
                          "GPU-RTX4080", 36, today, 16, "PCIe 4.0");
        HDD hdd = new HDD("SSD 1TB", "Samsung", 10000, 6,
                          "SSD-1TB", 60, today, 1000, 7200);

        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(gpu);
        pc.AddComponent(hdd);

        return pc;
    }
}

public class OfficeComputerBuilder : IComputerBuilder
{
    public Computer Build(string model)
    {
        Computer pc = new Computer(model);
        DateTime today = DateTime.Now;

        CPU cpu = new CPU("Core i5-12400", "Intel", 18000, 65,
                          "CPU-i5", 24, today, 6, 2.5, "LGA1700");
        RAM ram = new RAM("DDR4 16GB", "Crucial", 5000, 10,
                          "RAM-DDR4-16", 24, today, 16, 3200);
        HDD hdd = new HDD("SSD 512GB", "WD", 4500, 4,
                          "SSD-512", 36, today, 512, 7200);

        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(hdd);

        return pc;
    }
}

public class ComputerService
{
    private IComputerBuilder _builder;

    public ComputerService(IComputerBuilder builder)
    {
        if (builder == null)
        {
            throw new ArgumentNullException("builder");
        }
        _builder = builder;
    }

    public Computer Assemble(string model)
    {
        return _builder.Build(model);
    }
}

Console.WriteLine("=== Игровой ПК ===");
IComputerBuilder gamingBuilder = new GamingComputerBuilder();
ComputerService gamingService = new ComputerService(gamingBuilder);
Computer gaming = gamingService.Assemble("Gaming PC");
gaming.ShowAll();

Console.WriteLine("=== Офисный ПК ===");
IComputerBuilder officeBuilder = new OfficeComputerBuilder();
ComputerService officeService = new ComputerService(officeBuilder);
Computer office = officeService.Assemble("Office PC");
office.ShowAll();

Console.WriteLine("=== Репозиторий компонентов ===");
IComponentRepository repo = new InMemoryComponentRepository();
DateTime now = DateTime.Now;

CPU cpu1 = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, "SN1", 36, now, 8, 4.2, "AM5");
CPU cpu2 = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, "SN2", 36, now, 8, 4.2, "AM5");
CPU cpu3 = new CPU("Core i5-12400", "Intel", 18000, 65, "SN3", 24, now, 6, 2.5, "LGA1700");

repo.Add(cpu1);
repo.Add(cpu3);

Console.WriteLine("cpu1 == cpu2: " + (cpu1 == cpu2));
Console.WriteLine("cpu1 != cpu3: " + (cpu1 != cpu3));
Console.WriteLine("ToString: " + cpu1.ToString());

Component found = repo.FindByName("Core i5-12400");
if (found != null)
{
    Console.WriteLine("Найден: " + found.Name);
}
else
{
    Console.WriteLine("Компонент не найден.");
}

bool removed = repo.Remove(cpu2);
if (removed == true)
{
    Console.WriteLine("Компонент cpu2 удалён из репозитория.");
}
else
{
    Console.WriteLine("Компонент cpu2 не найден (но равен cpu1 по Equals — удаление возможно).");
}

int count = 0;
foreach (Component c in repo.GetAll())
{
    count += 1;
}
Console.WriteLine("Всего в репозитории: " + count);
```
