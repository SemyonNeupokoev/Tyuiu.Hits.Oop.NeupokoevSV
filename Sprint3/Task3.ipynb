```
public abstract class Component
{
    private string _name;
    private string _manufacturer;
    private decimal _price;
    private int _powerConsumption;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = value; }
    }

    public decimal Price
    {
        get { return _price; }
        set { _price = value; }
    }

    public int PowerConsumption
    {
        get { return _powerConsumption; }
        set { _powerConsumption = value; }
    }

    protected Component(string name, string manufacturer, decimal price, int powerConsumption)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
        PowerConsumption = powerConsumption;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Производитель: {Manufacturer}, Цена: {Price}, Потребление: {PowerConsumption} Вт");
    }

    public override string ToString()
    {
        return $"{Name} ({Manufacturer}) - {Price} руб.";
    }

    public static bool operator ==(Component a, Component b)
    {
        if (a is null && b is null) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(Component a, Component b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is Component c && this == c;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price);
    }
}

public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable
{
    private int _cores;
    private double _frequency;

    public int Cores
    {
        get { return _cores; }
        set { _cores = value; }
    }

    public double Frequency
    {
        get { return _frequency; }
        set { _frequency = value; }
    }

    public CPU(string name, string manufacturer, decimal price, int powerConsumption, int cores, double frequency)
        : base(name, manufacturer, price, powerConsumption)
    {
        Cores = cores;
        Frequency = frequency;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Ядер: {Cores}, Частота: {Frequency} ГГц");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен в сокет.");
    }
}

public class RAM : Component, IInstallable
{
    private int _capacityGB;
    private int _frequencyMHz;

    public int CapacityGB
    {
        get { return _capacityGB; }
        set { _capacityGB = value; }
    }

    public int FrequencyMHz
    {
        get { return _frequencyMHz; }
        set { _frequencyMHz = value; }
    }

    public RAM(string name, string manufacturer, decimal price, int powerConsumption, int capacityGB, int frequencyMHz)
        : base(name, manufacturer, price, powerConsumption)
    {
        CapacityGB = capacityGB;
        FrequencyMHz = frequencyMHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объём: {CapacityGB} ГБ, Частота: {FrequencyMHz} МГц");
    }

    public void Install()
    {
        Console.WriteLine($"Память {Name} установлена в слот DIMM.");
    }
}

public class GPU : Component, IInstallable
{
    private int _videoMemoryGB;
    private string _interface;

    public int VideoMemoryGB
    {
        get { return _videoMemoryGB; }
        set { _videoMemoryGB = value; }
    }

    public string Interface
    {
        get { return _interface; }
        set { _interface = value; }
    }

    public GPU(string name, string manufacturer, decimal price, int powerConsumption, int videoMemoryGB, string @interface)
        : base(name, manufacturer, price, powerConsumption)
    {
        VideoMemoryGB = videoMemoryGB;
        Interface = @interface;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Видеопамять: {VideoMemoryGB} ГБ, Интерфейс: {Interface}");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена в слот {Interface}.");
    }
}

public class HDD : Component, IInstallable
{
    private int _capacityGB;
    private int _rpm;

    public int CapacityGB
    {
        get { return _capacityGB; }
        set { _capacityGB = value; }
    }

    public int Rpm
    {
        get { return _rpm; }
        set { _rpm = value; }
    }

    public HDD(string name, string manufacturer, decimal price, int powerConsumption, int capacityGB, int rpm)
        : base(name, manufacturer, price, powerConsumption)
    {
        CapacityGB = capacityGB;
        Rpm = rpm;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объём: {CapacityGB} ГБ, Скорость: {Rpm} об/мин");
    }

    public void Install()
    {
        Console.WriteLine($"Диск {Name} установлен в отсек 3.5\".");
    }
}

public class Computer
{
    private string _model;
    private List<Component> _components = new List<Component>();

    public string Model
    {
        get { return _model; }
        set { _model = value; }
    }

    public Computer(string model)
    {
        Model = model;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Name} добавлен в {Model}.");
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удалён из {Model}.");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден в {Model}.");
        }
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (Component c in _components)
        {
            total += c.Price;
        }
        return total;
    }

    public void ShowAll()
    {
        Console.WriteLine($"--- Компьютер {Model} ({_components.Count} компонентов) ---");
        foreach (Component c in _components)
        {
            c.GetInfo();
            Console.WriteLine();
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.");
    }
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        Computer pc = new Computer("Gaming PC");
        pc.AddComponent(new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, 8, 4.2));
        pc.AddComponent(new RAM("DDR5 32GB", "Kingston", 12000, 15, 32, 6000));
        pc.AddComponent(new GPU("RTX 4080", "NVIDIA", 120000, 320, 16, "PCIe 4.0"));
        pc.AddComponent(new HDD("SSD 1TB", "Samsung", 10000, 6, 1000, 7200));
        return pc;
    }

    public static Computer CreateOfficePC()
    {
        Computer pc = new Computer("Office PC");
        pc.AddComponent(new CPU("Core i5-12400", "Intel", 18000, 65, 6, 2.5));
        pc.AddComponent(new RAM("DDR4 16GB", "Crucial", 5000, 10, 16, 3200));
        pc.AddComponent(new HDD("SSD 512GB", "WD", 4500, 4, 512, 7200));
        return pc;
    }
}

Console.WriteLine("Создание игрового ПК");
Computer gaming = ComputerFactory.CreateGamingPC();
Console.WriteLine();
gaming.ShowAll();

Console.WriteLine("Создание офисного ПК");
Computer office = ComputerFactory.CreateOfficePC();
Console.WriteLine();
office.ShowAll();

Console.WriteLine("Добавление и удаление компонента");
CPU extraCpu = new CPU("Core i9-14900K", "Intel", 60000, 253, 24, 3.2);
office.AddComponent(extraCpu);
office.RemoveComponent(extraCpu);

Console.WriteLine("Сравнение компонентов");
CPU cpu1 = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, 8, 4.2);
CPU cpu2 = new CPU("Ryzen 7 7800X3D", "AMD", 45000, 120, 8, 4.2);
CPU cpu3 = new CPU("Core i5-12400", "Intel", 18000, 65, 6, 2.5);

Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 != cpu3: {cpu1 != cpu3}");
Console.WriteLine($"cpu1.ToString(): {cpu1}");
```

```
public class Subscription
{
    private int _subscriptionId;
    private string _serviceName;
    private decimal _cost;
    private bool _isActive;

    public int SubscriptionId
    {
        get { return _subscriptionId; }
        set { _subscriptionId = value; }
    }

    public string ServiceName
    {
        get { return _serviceName; }
        set { _serviceName = value; }
    }

    public decimal Cost
    {
        get { return _cost; }
        set { _cost = value; }
    }

    public bool IsActive
    {
        get { return _isActive; }
        set { _isActive = value; }
    }

    public Subscription(int subscriptionId, string serviceName, decimal cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
        IsActive = true;
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Название: {ServiceName}, Стоимость: {Cost}, Активна: {IsActive}");
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual void ExtendSubscription(int months)
    {
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Стоимость: {Cost * months}");
    }

    public void Activate()
    {
        IsActive = true;
        Console.WriteLine($"Подписка {ServiceName} активирована.");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"Подписка {ServiceName} деактивирована.");
    }

    public void ExtendSubscription(int months, decimal bonus)
    {
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев с бонусом {bonus}");
        Console.WriteLine($"Стоимость: {Cost * months - bonus}");
    }

    public decimal CalculateMonthlyCost(int discountPercent)
    {
        return Cost * (1 - discountPercent / 100m);
    }
}

public class OnlineServiceSubscription : Subscription
{
    private int _maxUsers;
    private int _storageGB;
    private bool _hasSupport;

    public int MaxUsers
    {
        get { return _maxUsers; }
        set { _maxUsers = value; }
    }

    public int StorageGB
    {
        get { return _storageGB; }
        set { _storageGB = value; }
    }

    public bool HasSupport
    {
        get { return _hasSupport; }
        set { _hasSupport = value; }
    }

    public OnlineServiceSubscription(int subscriptionId, string serviceName, decimal cost, int maxUsers)
        : base(subscriptionId, serviceName, cost)
    {
        MaxUsers = maxUsers;
        StorageGB = 10;
        HasSupport = false;
    }

    public override decimal CalculateMonthlyCost()
    {
        return Cost + (MaxUsers - 1) * Cost * 0.1m;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Пользователей: {MaxUsers}, Хранилище: {StorageGB} ГБ, Поддержка: {HasSupport}");
    }

    public void UpgradeStorage(int extraGB)
    {
        StorageGB += extraGB;
        Console.WriteLine($"Хранилище увеличено на {extraGB} ГБ. Теперь: {StorageGB} ГБ");
    }

    public void ToggleSupport()
    {
        HasSupport = !HasSupport;
        Console.WriteLine($"Поддержка: {(HasSupport ? "включена" : "выключена")}");
    }

    public void UpgradeStorage(int extraGB, bool premium)
    {
        int bonus = premium ? extraGB / 2 : 0;
        StorageGB += extraGB + bonus;
        Console.WriteLine($"Хранилище увеличено на {extraGB} ГБ (+{bonus} бонус). Теперь: {StorageGB} ГБ");
    }
}

public class StreamingSubscription : Subscription
{
    private int _maxStreams;
    private string _videoQuality;
    private bool _hasAds;

    public int MaxStreams
    {
        get { return _maxStreams; }
        set { _maxStreams = value; }
    }

    public string VideoQuality
    {
        get { return _videoQuality; }
        set { _videoQuality = value; }
    }

    public bool HasAds
    {
        get { return _hasAds; }
        set { _hasAds = value; }
    }

    public StreamingSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost)
    {
        MaxStreams = maxStreams;
        VideoQuality = "HD";
        HasAds = true;
    }

    public override void ExtendSubscription(int months)
    {
        decimal discount = 0;
        if (months >= 12) discount = 0.2m;
        else if (months >= 6) discount = 0.1m;

        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяцев");
        Console.WriteLine($"Скидка: {discount * 100}%");
        Console.WriteLine($"Итоговая стоимость: {Cost * months * (1 - discount)}");
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Потоков: {MaxStreams}, Качество: {VideoQuality}, Реклама: {HasAds}");
    }

    public void UpgradeQuality(string newQuality)
    {
        VideoQuality = newQuality;
        Console.WriteLine($"Качество видео изменено на: {VideoQuality}");
    }

    public void RemoveAds()
    {
        HasAds = false;
        Console.WriteLine("Реклама отключена.");
    }

    public void UpgradeQuality(string newQuality, bool applyToAllDevices)
    {
        VideoQuality = newQuality;
        Console.WriteLine($"Качество видео изменено на: {VideoQuality} (на всех устройствах: {applyToAllDevices})");
    }
}

public class PremiumStreamingSubscription : StreamingSubscription
{
    private int _exclusiveContentCount;
    private bool _hasDolbyAtmos;

    public int ExclusiveContentCount
    {
        get { return _exclusiveContentCount; }
        set { _exclusiveContentCount = value; }
    }

    public bool HasDolbyAtmos
    {
        get { return _hasDolbyAtmos; }
        set { _hasDolbyAtmos = value; }
    }

    public PremiumStreamingSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost, maxStreams)
    {
        ExclusiveContentCount = 50;
        HasDolbyAtmos = true;
        VideoQuality = "4K";
        HasAds = false;
    }

    public override decimal CalculateMonthlyCost()
    {
        return base.CalculateMonthlyCost() + 300;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Эксклюзивного контента: {ExclusiveContentCount}, Dolby Atmos: {HasDolbyAtmos}");
    }

    public void AddExclusiveContent(int count)
    {
        ExclusiveContentCount += count;
        Console.WriteLine($"Добавлено {count} единиц контента. Всего: {ExclusiveContentCount}");
    }

    public void AddExclusiveContent(int count, string category)
    {
        ExclusiveContentCount += count;
        Console.WriteLine($"Добавлено {count} единиц контента в категорию «{category}». Всего: {ExclusiveContentCount}");
    }
}

public interface IDiscountable
{
    decimal ApplyDiscount(decimal amount, int percent);
}

public interface IGiftable
{
    void GiftTo(string recipient);
}

public class GiftablePremiumSubscription : PremiumStreamingSubscription, IDiscountable, IGiftable
{
    public GiftablePremiumSubscription(int subscriptionId, string serviceName, decimal cost, int maxStreams)
        : base(subscriptionId, serviceName, cost, maxStreams)
    {
    }

    public decimal ApplyDiscount(decimal amount, int percent)
    {
        decimal result = amount * (1 - percent / 100m);
        Console.WriteLine($"Применена скидка {percent}%: {amount} -> {result}");
        return result;
    }

    public void GiftTo(string recipient)
    {
        Console.WriteLine($"Подписка {ServiceName} подарена пользователю {recipient}");
    }
}

public class User
{
    private string _name;
    private decimal _balance;
    private bool _isPremium;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public decimal Balance
    {
        get { return _balance; }
        set { _balance = value; }
    }

    public bool IsPremium
    {
        get { return _isPremium; }
        set { _isPremium = value; }
    }

    public User(string name)
    {
        Name = name;
        Balance = 0;
        IsPremium = false;
    }

    public void ShowSubscription(Subscription sub)
    {
        Console.WriteLine($"Пользователь {Name} смотрит подписку:");
        sub.GetSubscriptionDetails();
        Console.WriteLine($"Ежемесячная стоимость: {sub.CalculateMonthlyCost()}");
    }

    public void Deposit(decimal amount)
    {
        Balance += amount;
        Console.WriteLine($"{Name} пополнил баланс на {amount}. Баланс: {Balance}");
    }

    public bool Pay(decimal amount)
    {
        if (Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"{Name} оплатил {amount}. Остаток: {Balance}");
            return true;
        }
        Console.WriteLine($"Недостаточно средств у {Name}. Нужно: {amount}, есть: {Balance}");
        return false;
    }

    public void ShowBalance()
    {
        Console.WriteLine($"Баланс {Name}: {Balance}, Премиум: {IsPremium}");
    }

    public void Deposit(decimal amount, bool bonus)
    {
        decimal extra = bonus ? amount * 0.1m : 0;
        Balance += amount + extra;
        Console.WriteLine($"{Name} пополнил баланс на {amount} (+{extra} бонус). Баланс: {Balance}");
    }
}

public class SubscriptionBox<T> where T : Subscription
{
    private List<T> _items = new List<T>();

    public void Add(T item)
    {
        _items.Add(item);
        Console.WriteLine($"Добавлена подписка: {item.ServiceName}");
    }

    public void ShowAll()
    {
        Console.WriteLine($"--- Содержимое коробки ({_items.Count} шт.) ---");
        foreach (T item in _items)
        {
            item.GetSubscriptionDetails();
            Console.WriteLine($"Стоимость: {item.CalculateMonthlyCost()}");
            Console.WriteLine();
        }
    }

    public decimal TotalCost()
    {
        decimal total = 0;
        foreach (T item in _items)
        {
            total += item.CalculateMonthlyCost();
        }
        return total;
    }

    public T GetById(int id)
    {
        foreach (T item in _items)
        {
            if (item.SubscriptionId == id) return item;
        }
        return null;
    }
}

User user = new User("Иван");
user.Deposit(2000);

OnlineServiceSubscription online = new OnlineServiceSubscription(1, "Microsoft 365", 500, 5);
StreamingSubscription stream = new StreamingSubscription(2, "Netflix", 899, 4);

Console.WriteLine("=== Простое наследование ===");
user.ShowSubscription(online);
Console.WriteLine();
user.ShowSubscription(stream);

Console.WriteLine("\n=== Сложное наследование ===");
PremiumStreamingSubscription premium = new PremiumStreamingSubscription(3, "Netflix Premium", 1299, 6);
user.ShowSubscription(premium);
premium.AddExclusiveContent(10);

Console.WriteLine("\n=== Множественное наследование ===");
GiftablePremiumSubscription gift = new GiftablePremiumSubscription(4, "Netflix Gift", 1499, 6);
user.ShowSubscription(gift);
gift.GiftTo("Мария");
gift.ApplyDiscount(1499, 15);

Console.WriteLine("\n=== Перегрузка методов ===");
online.UpgradeStorage(20);
online.UpgradeStorage(20, true);
stream.UpgradeQuality("4K");
stream.UpgradeQuality("4K", true);
premium.AddExclusiveContent(10);
premium.AddExclusiveContent(10, "Спорт");
user.Deposit(500);
user.Deposit(500, true);
stream.ExtendSubscription(12);
stream.ExtendSubscription(12, 200);

Console.WriteLine("\n=== Generic-класс ===");
SubscriptionBox<StreamingSubscription> streamBox = new SubscriptionBox<StreamingSubscription>();
streamBox.Add(stream);
streamBox.Add(premium);
streamBox.Add(gift);
streamBox.ShowAll();
Console.WriteLine($"Итого по коробке: {streamBox.TotalCost()}");
Console.WriteLine($"Поиск по ID 3: {streamBox.GetById(3).ServiceName}");

SubscriptionBox<Subscription> allBox = new SubscriptionBox<Subscription>();
allBox.Add(online);
allBox.Add(stream);
allBox.Add(premium);
allBox.Add(gift);
Console.WriteLine($"\nИтого по всем подпискам: {allBox.TotalCost()}");

Console.WriteLine("\n=== Оплата ===");
user.Pay(1200);
user.ShowBalance();
```
